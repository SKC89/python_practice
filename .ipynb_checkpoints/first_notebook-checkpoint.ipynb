```python
def square(n):
    return n*n
```

```python
square(7)
```

```python
print(square(2))
print(square(3))
```

```python
temps = [18, 25, 31, 12, 27]
count=0
for x in temps:
    if x>20:
        count=count+1
print(count)
```

```python
def average(nums):
    total=sum(nums)
    return total/len(nums)
print(average(temps))
```

```python
total=0
for x in temps:
    total=total+x
average=total/len(temps)
print(average)
```
